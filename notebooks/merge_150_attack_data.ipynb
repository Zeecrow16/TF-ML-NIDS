```python
import pandas as pd
```

```python
df_cwt = pd.read_csv("data/cwt_150_multiclass_attack.csv")
df_unsw = pd.read_csv("../data/NF-UNSW-NB15-v3.csv")
```

```python
df_unsw['flow_id'] = df_unsw.index
```

```python
get_temporal = [
    "FLOW_DURATION_MILLISECONDS",
    "SRC_TO_DST_IAT_MIN",
    "SRC_TO_DST_IAT_MAX",
    "SRC_TO_DST_IAT_AVG",
    "SRC_TO_DST_IAT_STDDEV", 
    "DST_TO_SRC_IAT_MIN",
    "DST_TO_SRC_IAT_MAX",
    "DST_TO_SRC_IAT_AVG",
    "DST_TO_SRC_IAT_STDDEV",
    "flow_id"
]
```

```python
df_temporal = df_unsw[get_temporal]
```

```python
df_merged = pd.merge(df_cwt, df_temporal, on="flow_id", how="inner")
```

```python
df_merged.to_csv("../data/ml_150_multiclass_data.csv", index=False)
```
